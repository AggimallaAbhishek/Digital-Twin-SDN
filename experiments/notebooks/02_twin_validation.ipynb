```python
import json
from pathlib import Path

report = json.loads(Path('../../models/twin/v1/validation.json').read_text())
print('datasets:', report['datasets'], '| commit:', report['git_commit'])
print('bulk offered rate (Mbit/s):', report['bulk_offered_mbps'])
```

```python
for split, table in report['after_calibration'].items():
    print(f'--- {split}')
    for key, row in table.items():
        mape = 'n/a' if row['mape'] is None else f"{row['mape']:.1%}"
        print(f'{key:28s} MAPE {mape:>7s}  n={row["n"]}')
```
